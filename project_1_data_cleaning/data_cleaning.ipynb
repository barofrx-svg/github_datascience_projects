# 2. Cleaning 

### Column names
Since column names contain typos and spaces, this is the first obstacle I solved. I renamed the columns in a way that lets me query them easily. (Using dashes)

```python
df = df.rename(columns={"IMBD title ID": "Title",
                        "Original titlÊ" : "Original_title",
                        "Release year": "Release_year",
                        "Genrë¨" : "Genre",
                        "Content Rating": "Content_rating",
                        })
```

```python
df.columns
```

### Removing noise in columns
One column did not have any significant meaning, so I removed it from the dataframe.

```python
df = df.drop(columns=["Unnamed: 8"])
df
```

### Overviewing data integrity

#### I. Title

- Every record in column "Title" mathces the pattern of 2 letters followed by 7 digits.
- Only one exception, this record contains NaN in every column, so I dropped it later.

```python
# Title
df[df["Title"].str.contains(r"[a-z]{2}\d{7}")]
```

To further narrow the pattern, I also noticed that the title always starts with tt. No other words combination were visible.

```python
# Every row in title starts with tt then 7 digits
df[df["Title"].str.contains(r"tt\d{7}")]
```

Here I dropped the row that does not contain any information.

```python
df[~df["Title"].str.contains(r"[a-z]{2}\d{7}")]
```

```python
df = df.drop(index=13).reset_index(drop=True)
len(df)
```

#### II. Release year
Since release year had multiple formats, I united them into a simple one. Most of the dates followed the yyyy-mm-dd format so I chose this as my basis. Then I extraced from every row the year, month and day and formulated into yyyy-mm-dd.

```python
# Most of the date has the format yyyy-mm-dd, so extract them and then reunite later
df[df["Release_year"].str.contains(r"\d{4}-\d{2}-\d{2}")]
```

```python
# These does not follow the format, 9 cases
# since almost all of them are different the easieast is to alter them manually
df[~df["Release_year"].str.contains(r"\d{4}-\d{2}-\d{2}")]
```

```python
df.iloc[1,2] = "1972-09-21"
df.iloc[2,2] = "2008-07-23"
df.iloc[5,2] = "2022-02-04"
df.iloc[9,2] = "1999-10-29"
df.iloc[12,2] = "1966-12-23"
df.iloc[14,2] = "2003-01-16"
df.iloc[17,2] = "1976-11-18"
df.iloc[44,2] = "1946-11-21"
df.iloc[69,2] = "1951-03-06"
df[~df["Release_year"].str.contains(r"\d{4}-\d{2}-\d{2}")]
# everything follows the rule above, now change the data type to date
```

```python
df["Release_year"] = df["Release_year"].str.replace("-", ".")
df
```

#### III. Moving on to genre

```python
# It follows the rule that only contain upper, lower case letters and comma
df[~df["Genre"].str.contains(r"[a-z|,|A-Z]+")]
```

```python
# made sure the rule is strict, it contains only letters.
df[df["Genre"].str.contains(r"\d")]
```

#### IV. Duration

First I checked the records that contain something else other than digits

```python
df[df["Duration"].str.contains(r"\D")]
```

Then I extracted the digits from the column and dropped the letters or dashes, dots etc.

```python
df["Duration"] = df["Duration"].str.extract(r"(\d+)")
df
```

After this the column is united. In this column every record contains only digits.

```python
df[df["Duration"].str.contains(r"\D")]
```

Records that contain NaN should display the "No duration" class, however since it is a numerical column, adding this leads to future errors during the data analysis phase. The best way is to just leave them as NaN.

#### V. Country

First I checked the list and the counts of each country, and immidietly noticed that there are typos, generatig more classes. So regarding this column, I united the countries. USA and US. are the same. And did this with the rest of the countries

```python
df["Country"].value_counts()
```

- Here I replaced everything with USA that starts with US and ends some other way than A.
- Formulated logic for the rest of the countries as well.

```python
df["Country"] = df["Country"].str.replace(r"^US\S?", "USA", regex=True) 
df["Country"] = df["Country"].str.replace(r"^(N|n)(w|W|ew) ?[z|Z]\w+|\w?[d|D]$", "New Zealand", regex=True)
df["Country"] = df["Country"].str.replace(r"\w?[i|I][taly]*[y|Y|t|T]\w?", "Italy", regex=True)
```

And resulted in an united list

```python
df["Country"].value_counts()
```

#### VI. Content rating

First listed the classes

```python
df["Content_rating"].value_counts()
```

```python
#get indexes since its easier
not_rated_mask = ~df["Content_rating"].str.contains(r"^R|^PG-13|^PG|^G", na=False)

# replace those that does not follow the MPAA age restrictions with not rated.
df[~df["Content_rating"].str.contains(r"^R|^PG-13|^PG|^G")]

# the following list shows the records where content rating does not follow R, PG-13, PR or G rating
```

```python
# Replace these records with "not rated" to unite them into a single group
df.loc[not_rated_mask, "Content_rating"] = "Not rated"
```

```python
df["Content_rating"].value_counts()
```

#### VII. Income

```python
df
```

```python
# records that contains letters
df[df["Income"].str.contains(r"[a-zA-Z]")]
```

Since there is only one case where letter is included it is easier to just manually replace it, than writing a sorting algorithm

```python
# replace o with 0
df["Income"] = df["Income"].str.replace("o", "0")
```

```python
# every row contains the $? -> yes
df[df["Income"].str.contains(r"\$")]
```

```python
# remove dollar and white space
df["Income"] = df["Income"].str.replace(r"\$", "", regex=True)
df["Income"] = df["Income"].str.replace(r"\s", "", regex=True)
```

```python
df
```

```python
# does the record only contain numbers or other characters like comma or dot aswell?
# well, there is one row that does not follow this rule, it contains commas
df[~df["Income"].str.contains(r"^[0-9]+$")]
```

```python
# just remove the commas so the records follow a single rule
df["Income"] = df["Income"].str.replace(",", "")
df
```

Add dots every 3 digits to read number off easily.

```python
# make sure records are numeric and not text
df["Income"] = pd.to_numeric(df["Income"], errors='coerce')
df
```

#### VIII. Votes

```python
# votes column had an invisible " " white space
df.columns = df.columns.str.strip()
```

```python
# every number contains the dot as separator, lets remove it
df[~df["Votes"].str.contains(r"\.")]
```

```python
df
```

#### IX. Score

First I looked at records that does not only contain digits.

```python
df[~df["Score"].str.contains(r"^\d\.\d$")]
```

```python
# replace these, that do not follow the d.d rule
df["Score"] = df["Score"].str.replace(",.", ".")
df["Score"] = df["Score"].str.replace(r"\[a-zA-Z]", "", regex=True)
df["Score"] = df["Score"].str.replace(",", ".")
df["Score"] = df["Score"].str.replace("..", ".")
df["Score"] = df["Score"].str.replace(":", ".")
df["Score"] = df["Score"].str.replace(r"^9\.$", "9.0", regex=True)
df["Score"] = df["Score"].str.replace("++", "")
```

```python
# do not write new rules, just extract the d.d values
df[~df["Score"].str.contains(r"^\d\.\d$")]
```

```python
df["Score"] = df["Score"].str.extract(r"(\d\.\d)")
```

```python
# done
df[~df["Score"].str.contains(r"^\d\.\d$")]
```

#### X. Original title

```python
# every title follows the rule: letters and numbers only
df[~df["Original_title"].str.contains(r"\w+")]
```

```python
# these values contain white space, need to strip the entries
df[df["Original_title"].str.contains(r"\s+")]
```

```python
# start and ends with white space, none so this is done
df[df["Original_title"].str.contains(r"(^\s\w+\s$)|(^\s\w+)|(\w+\s$)", regex=True)]
```

#### XI. Director

```python
# every director name follows that it only contains letters
df[~df["Director"].str.contains(r"[a-zA-Z]")]
```

```python
# now look at the free spaces
# sweet, no unwanted free space
df[df["Director"].str.contains(r"(^\s\w+\s$)|(^\s\w+)|(\w+\s$)", regex=True)]
```

# 3. Final polishing

Adding $ to income column so numbers have meaning.

```python
df = df.rename(columns={"Income": "Income_$"})
```

Then I converted every column into the appropriate data type.

```python
# converting into numbers
df["Income_$"] = pd.to_numeric(df["Income_$"], errors='coerce')
df["Duration"] = pd.to_numeric(df["Duration"], errors='coerce')
df["Votes"] = df["Votes"].str.replace(".", "")
df["Votes"] = pd.to_numeric(df["Votes"], errors='coerce')
df["Score"] = pd.to_numeric(df["Score"], errors='coerce')

# converting into date
df["Release_year"] = pd.to_datetime(df["Release_year"], errors='coerce')

df
```

And created the final, cleaned dataframe.

```python
final_df = df.copy()
```
